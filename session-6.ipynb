# Trigonometric Functions
#This notebook defines and uses functions for the sine and cosine of an angle.

#defnes and returns the sine of x.

```python
import math

def calculate_sin(x):
    return math.sin(x)
```

#defnes and returns the cos of x.

```python
import math

def cosine(x):
     return math.cos(x)
```

# Generate x values from 0 to 2 with a thousand entries
# Calculate sin(x) and cos(x) using the defined functions
# Create a Pandas DataFrame for tabulation

```python
import numpy as np
import pandas as pd

x_values = np.linspace(0, 2, 1000)

sin_values = [calculate_sin(val) for val in x_values]
cos_values = [calculate_cos(val) for val in x_values]

tabulated_data = pd.DataFrame({
    'x': x_values,
    'sin(x)': sin_values,
    'cos(x)': cos_values
})

print("Tabulated values for sin(x) and cos(x) vs. x:")
display(tabulated_data.head())
display(tabulated_data.tail())
```

```python
print("First 10 values of x, sin(x), and cos(x) in columns:")
print(f"{'x':<10} {'sin(x)':<10} {'cos(x)':<10}")
print("-"*30)
for i in range(10):
    x_val = tabulated_data.iloc[i]['x']
    sin_val = tabulated_data.iloc[i]['sin(x)']
    cos_val = tabulated_data.iloc[i]['cos(x)']
    print(f"{x_val:<10.6f} {sin_val:<10.6f} {cos_val:<10.6f}")
```
